```python
import os
import numpy as np
import scipy.io as sio
import h5py

# Define the directory containing the .mat files
directory = r'C:\Users\msa583\Documents\New Posture Detections - Part 2\82556\day2'

# List of .mat filenames to process
mat_files = [
    "trimmed_GyroX.mat", "trimmed_GyroY.mat", "trimmed_GyroZ.mat",
    "trimmed_MXR.mat", "trimmed_MYR.mat", "trimmed_MZR.mat"
]

# Dictionary to store the loaded data
data_dict = {}

# Loop through each .mat file to load and store the data
for mat_file in mat_files:
    # Construct full file path
    mat_path = os.path.join(directory, mat_file)
    
    # Load the .mat file
    mat_data = sio.loadmat(mat_path)
    
    # Extract the variable name (assuming the variable name is the same as the file name without extension)
    variable_name = mat_file.split('.')[0]

    # Extract the data from the mat file
    data = mat_data[variable_name]

    # Store the data in the dictionary
    data_dict[variable_name] = data

    # Print the reference of each signal
    print(f"{variable_name} corresponds to signal from {mat_file}")

# Construct the h5 file path
h5_path = os.path.join(directory, 'combined_data.h5')

# Save the data into a single .h5 file with separate datasets
with h5py.File(h5_path, 'w') as h5f:
    for key, data in data_dict.items():
        h5f.create_dataset(key, data=data)

# Function to recursively print the structure of the HDF5 file
def print_h5_structure(name, obj):
    if isinstance(obj, h5py.Group):
        print(f"Group: {name}")
    elif isinstance(obj, h5py.Dataset):
        print(f"  Dataset: {name}, shape: {obj.shape}, dtype: {obj.dtype}")

# Open the file to read and print its structure
with h5py.File(h5_path, 'r') as h5f:
    print("\nHDF5 file structure:")
    h5f.visititems(print_h5_structure)
```
